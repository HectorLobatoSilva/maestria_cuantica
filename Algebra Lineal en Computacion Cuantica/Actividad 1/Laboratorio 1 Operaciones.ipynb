```python
# imports de paquetes y funciones comunes
import numpy as np
import cmath
import math
from IPython.display import display, Math

def formato_binomial(z):
    """
    Convierte un número complejo z en una cadena formateada limpia para LaTeX.
    Detecta automáticamente si los componentes son enteros o flotantes.
    """
    real = round(z.real, 2)
    imag = round(z.imag, 2)
    
    def limpiar_numero(val):
        """Convierte float a str: '3.0' -> '3', '3.20' -> '3.2', '3.25' -> '3.25'"""
        if val.is_integer():
            return str(int(val))
        return f"{val:.2f}".rstrip('0').rstrip('.')

    str_real = limpiar_numero(real)
    str_imag_abs = limpiar_numero(abs(imag))
    
    # Caso 1: Es un número totalmente real (imag es 0)
    if imag == 0:
        return str_real
    
    # Caso 2: Es un número imaginario puro (real es 0)
    if real == 0:
        if imag == 1: return "i"
        if imag == -1: return "-i"
        return f"{limpiar_numero(imag)}i"
    
    # Caso 3: Número complejo completo (real + imaginario)
    signo = "+" if imag > 0 else "-"
    str_imag = "i" if str_imag_abs == "1" else f"{str_imag_abs}i"
    
    return f"{str_real} {signo} {str_imag}"

def formato_polar(z):
    r, theta = cmath.polar(z)
    return f"{r:.4f} \\cdot e^{{{theta:.4f}i}}"

def formato_matriz(matriz):
    """
    Convierte un ndarray de NumPy a código LaTeX de matriz automáticamente.
    Uso: display(Math(matriz_a_latex(M)))
    """
    cuerpo = r" \\ ".join(
        " & ".join(formato_binomial(elem) for elem in fila)
        for fila in matriz
    )
    return rf"\begin{{pmatrix}} {cuerpo} \end{{pmatrix}}"

def formato_vector(vector):
    """
    Convierte un ndarray 1D de NumPy o Statevector de Qiskit a un vector columna LaTeX.
    """
    vector_array = np.array(vector).flatten()
    cuerpo = r" \\ ".join(formato_binomial(elemento) for elemento in vector_array)
    return rf"\begin{{bmatrix}} { cuerpo } \end{{bmatrix}}"
```

## Considerando las siguientes matrices
- $A = [3+2i , 1-i , 4-3i , 2+i , -5+4i , 6 , -3+2i , 7-i , 8+3i]$ (9 elementos en total)
- $B = [2+3i , 4 , 1+2i , 3-2i , 5+i , 7-3i , 6+4i , -2 , 9+5i]$ (9 elementos en total)
- $C = [1+4i , 2-2i , 3+i , -5 , -2+3i , 6 , 4+7i , 2+i , -1+2i , -3-4i , -6+5i , 3-i , -8i , 7+4i , -5-3i , 5i]$ (16 elementos en total)
- $D = [i , 2 , 2-3i , 4+5i]$ (4 elementos en total)

Para que una matriz sea cuadrada ( estandar en álgebra lineal para poder calcular **determinantes, trazas, inversas y valores propios**)

- Matrices **$A \text{ y } B$** tienen **9 elementos**: La unica forma de acomodarlos en una matriz cuadrada es $3 \text{ x } 3(\sqrt(9) = 3)$
- Matriz $C$ tiene **16 elementos**: La unica forma de acomodarlos en una matriz cuadratica es en $4 \text{ x } 4 (\sqrt{16} = 4)$
- Matriz $D$ tiene **4 elementos**: La única forma de acomodar 4 elementes en una matriz cuadratica es en un arreglo de $2 \text { x } 2 (\sqrt{4} = 2)$

```python
# Como recordatorio, en python los números imaginarios i se escriben como j en el numero complejo
A = np.array([
    [  3 + 2j,  1 - 1j, 4 - 3j ],
    [  2 + 1j, -5 + 4j, 6 + 0j ],
    [ -3 + 2j,  7 - 1j, 8 + 3j ]
])

B = np.array([
    [ 2 + 3j,  4 + 0j, 1 + 2j ],
    [ 3 - 2j,  5 + 1j, 7 - 3j ],
    [ 6 + 4j, -2 + 0j, 9 + 5j ]
])

C = np.array([
    [  1 + 4j,  2 - 2j,  3 + 1j, -5 + 0j ],
    [ -2 + 3j,  6 + 0j,  4 + 7j,  2 + 1j ],
    [ -1 + 2j, -3 - 4j, -6 + 5j,  3 - 1j ],
    [  0 - 8j,  7 + 4j, -5 - 3j,  0 + 5j ]
])

D = np.array([
    [ 0 + 1j, 2 + 0j ],
    [ 2 - 3j, 4 + 5j ]
])

display(Math(rf"""
\begin{{aligned}} 
A &= {formato_matriz(A)} \\[1em]
B &= {formato_matriz(B)} \\[1em]
C &= {formato_matriz(C)} \\[1em]
D &= {formato_matriz(D)}
\end{{aligned}}
"""))
```

## La forma polar de cada uno de los determinantes y de la traza de cada matriz.

Un número complejo en su representación rectangular $z = a + bi$ se puede traducir a su forma polar $z = r e^{i\theta}$, donde $r = \vert{z}\vert = \sqrt{a^2 + b^2}$ representa la magnitud/módulo y $\theta = \text{atan2}(b, a)$ es el argumento o fase en radianes. En álgebra lineal sobre $\mathbb{C}$, la traza $\text{tr}(M) = \sum M_{ii}$ representa la suma de los valores propios, mientras que el determinante $\det(M) = \prod \lambda_i$ representa su producto.

### En resumen
1. **Traza ($\text{tr}$)**: Es la suma de los elementos de la diagonal principal de la matriz.
2. **Determinante ($\det$)**: Un escalar único que caracteriza las propiedades algebraicas de la matriz.
3. **Forma polar de un número complejo**: Todo numero complejo $z = a + bi$ en el plano cartesiano se puede escribir en coordenadas polares como:
$z = a + bi = \vert{z}\vert^{i\theta} = r e ^{i \theta } = r (\cos \theta + i \sin \theta)$
- $r = \vert{z}\vert = \sqrt{a^2 + b^2}$ es el **módulo** (la longitud del vector).
- $\theta = Arg(z)$ es la **fase o argumento**(el ángulo en radianes con el eje real).

#### Fuentes
- **Strang, G. (2016)**. Introduction to Linear Algebra (5th ed.). Wellesley-Cambridge Press. (Capítulos 6 y 9).


```python
matrices = { 'A': A, 'B': B, 'C': C, 'D': D }

def display_r(a,b):
    a_2 = a**2
    b_2 = b**2
    r =  math.sqrt(a_2 + b_2)
    return rf"r &= \sqrt{{({a:.0f}^2) + ({b:.0f}^2)}} = \sqrt{{{a_2:.0f} + {b_2:.0f}}} = \sqrt{{{(a_2 + b_2):.0f}}} \approx {r:.4f} \\"

def display_theta(a,b, tipo="det"):
    if tipo == "det":
        theta =  math.atan2(b,a)
        texto = rf"\operatorname{{arctan2}}({b:.0f},{a:.0f})"
    elif tipo == "tar":
        theta = math.atan(b/a)
        texto = rf"\operatorname{{arctan}}\left(\frac{{{b:.0f}}}{{{a:.0f}}}\right)"
    return rf"\theta &= {texto} \approx {theta:.4f} \text{{rad}} \\"

for matriz, valores in matrices.items():
    determinante = np.linalg.det(valores)
    traza = np.trace(valores)
    
    det_real, det_imag = round(determinante.real, 2), round(determinante.imag, 2)
    
    tr_real, tr_imag = round(traza.real, 2), round(traza.imag, 2)
    
    display(Math(rf"""
    \begin{{aligned}}
    \mathbf{{\text{{Matriz }} {matriz}}} &: \\
    \det({matriz}) &: \\
    {display_r(det_real, det_imag)}
    {display_theta(det_real, det_imag, tipo="det")}
    \det({matriz}) &= {formato_binomial(determinante)} \implies \mathbf{{{formato_polar(determinante)}}} \\
    \text{{tr}}({matriz}) &: \\
    {display_r(tr_real, tr_imag)}
    {display_theta(tr_real, tr_imag, tipo="tar")}
    \text{{tr}}({matriz}) &= {formato_binomial(traza)} \implies \mathbf{{{formato_polar(traza)}}}
    \end{{aligned}}
    """))
    print()
```

## $2i(det(D)+tr(C))AB-(1+i)  det⁡(C)tr(D)BA$

La estructura de la formula esta dividida en 2 terminos

$$\underbrace{2i(\det(D) + \text{tr}(C))}_{\mathbf{\text{Primer escalar } (k_1)}} \mathbf{AB} \quad - \quad \underbrace{(1+i)\det(C)\text{tr}(D)}_{\mathbf{\text{Segundo escalar } (k_2)}} \mathbf{BA}$$

$$k_1 = 2i \cdot (\det(D) + \text{tr}(C))$$

$$k_2 = (1+i) \cdot \det(C) \cdot \text{tr}(D)$$

Vamos a calcular las determinantes ($\det(D)$ y $\det(C)$), asi mismo la traza ($\text{tr(D)}$ y $\text{tr(C)}$) de estas matrices

```python
det_D = np.linalg.det(D)
tr_D = np.trace(D)
det_C = np.linalg.det(C)
tr_C = np.trace(C)

k1_parentesis = det_D + tr_C
k2_parentesis = det_C * tr_D

k1 = 2j * k1_parentesis
k2 = (1 + 1j)* k2_parentesis

AB = A @ B
BA = B @ A

resultado_matriz = ( k1 * AB ) - (k2 * BA )

display(Math(rf"""
\begin{{aligned}}
\text{{Calculando el escalar }} k_1 &= 2i \cdot (\det(D) + \text{{tr}}(C)) \\[0.2em]
\det(D) &= {formato_binomial(det_D)} \\[0.2em]
\text{{tr}}(C) &= {formato_binomial(tr_C)} \\[0.2em]
\det(D) + \text{{tr}}(C) &= ({formato_binomial(det_D)}) + ({formato_binomial(tr_C)}) = 
({formato_binomial(k1_parentesis)}) \\[0.2em]
\text{{Multiplicamos por }} 2i: \\[0.2em]
k_1 &= 2i \cdot ({formato_binomial(k1_parentesis)}) = \mathbf{{{formato_binomial(k1)}}} \\[1em]
\text{{Calculando el escalar }} k_2 &= (1 + i) \cdot \det(C) \cdot \text{{tr}}(D) \\[0.2em]
\det(C) &= {formato_binomial(det_C)} \\[0.2em]
\text{{tr}}(D) &= {formato_binomial(tr_D)} \\[0.2em]
(1 + i) \cdot \det(C) \cdot \text{{tr}}(D) &= (1 + i) \cdot ({formato_binomial(det_C)}) \cdot 
({formato_binomial(tr_D)}) = ({formato_binomial(k2_parentesis)}) \\[0.2em]
k_2 &= \mathbf{{{formato_binomial(k2)}}} \\[1em]
\mathbf{{\text{{Resultado Final:}}}} \\[0.2em]
k_1(AB) - k_2(BA) &= {formato_matriz(resultado_matriz)}
\end{{aligned}}
"""))
```

## Inversas de las matrices $A^{-1}, B^{-1}, C^{-1}, D^{-1}$

La **Inversa** de una matriz cuadrada $A$ se escribe como $A^{-1}$. Cuando existe, tiene la propiedad de que multiplicar $A$ por $A^{-1}$ da como resultado la matriz Identidad _$I$_( en cualquier orden ): $A^{-1} = I$ y $AA^{-1} = I$. La matriz A se denomina **invertible** o **no singular**.

Para que la matriz posea una inversa, debe cumplir de manera obligarotia con las siguientes condiciones fundaentales detalladas en el programa de álgebra lineal de Wellesley-Cambridge Press:
- **Ser una matriz cuadrada**: El número de filas debe ser exactamente igual al número de columnas ($nxn$). Las matrices rectangulares no tienen inversa bilateral estándar.
- **Determinante distinto a cero**: Su determinante debe cumplir con la condicion $\det{(A)} \neq 0$. Si el determinante es cero, la matriz se considera **singular** y no tiene inversa.
- **Columnas y filas linealmente independientes**: Ningula columna (o fila) puede ser una combinación lineal de las demás.
- **Solución única**: El sistema homogéneo $Ax = 0$ debe tener únicamente la solución trivial $(x = 0)$, lo que significa que el núcleo o _nullspace_ de la metriz contiene solo el vector cero.
- **Rango completo**: La matriz debe tener un rango igual a su dimensión $n$ (rango completo)

#### Fuente
- **Strang, G. (2016)**. Introduction to Linear Algebra (5th ed.). Wellesley-Cambridge Press. (Capítulo 2).

```python
texto = []
for matriz, valor in matrices.items():
    try:
        matriz_inversa = np.linalg.inv(valor)
        texto.append(rf"{matriz}^{{-1}} &= {formato_matriz(matriz_inversa)} \\[1em]")
    except np.linalg.LinAlgError:
        texto.append(rf"\text{{La matriz }} {matriz} &\text{{ no es invertible}} \\")

display(Math(rf"""
\begin{{aligned}}
{" ".join(texto)}
\end{{aligned}}
"""))
```

## Clasificación Formal de Matrices

Recordemos las diferentes tipos de matrices

### Matriz Hermintiana ($A = A^\dagger$)
Una matriz cuadrada cuyos elementos cimples que $a_ij = \overline a_ij$ (igual a su transpuesta conjugada). Una propiedad fundamental es que **todos sus valores propios ($\lambda$) estan obligados a ser números puramente reales**

### Matriz Unitaria ($U^\dagger U = I$)
Es una matriz compleja cuyas columnas son vectores oronormales entre si. Cumple con su inversa es exactamente igual a su transpuesta conjugada ($U^{-1} = U^H$)

### Matriz Normal ($AA^\dagger = A^\dagger A$)
Toda matriz Hermitiana y toda matriz Unitaria son por definición normales. Permiten la diagonalización mediante matrices unitarias.

#### Fuente
- **Strang, G. (2016)**. Introduction to Linear Algebra (5th ed.). Wellesley-Cambridge Press. (Capítulo 9).

```python
for matriz, valor in matrices.items():
    n = valor.shape[0]
    valor_conjugada_transpuesta = valor.conj().T
    I = np.eye(n)

    es_hermitiana = np.allclose(valor, valor_conjugada_transpuesta)
    es_unitaria = np.allclose(valor_conjugada_transpuesta @ valor, I)
    es_normal = np.allclose(valor @ valor_conjugada_transpuesta, valor_conjugada_transpuesta @ valor)

    valores_propios = np.linalg.eigvals(valor)

    display(Math(rf"""
    \begin{{aligned}}
    \mathbf{{\text{{Matriz }} {matriz}:}} &\\[0.3em]
    \text{{Valores propios }} \lambda &= \left[ {', '.join([formato_binomial(val) for val in valores_propios])} \right] \\[0.2em]
    \text{{¿Es Hermitiana? }} ({matriz} \stackrel{{?}}{{=}} {matriz}^\dagger) &\implies \mathbf{{{es_hermitiana}}} \\[0.2em]
    \text{{¿Es Unitaria? }} ({matriz}^\dagger {matriz} \stackrel{{?}}{{=}} I) &\implies \mathbf{{{es_unitaria}}} \\[0.2em]
    \text{{¿Es Normal? }} ({matriz}{matriz}^\dagger \stackrel{{?}}{{=}} {matriz}^\dagger {matriz}) &\implies \mathbf{{{es_normal}}}
    \end{{aligned}}
    """))
    print()
```

Las verificaciones para las matrices $A,B,C,D$ devolvieron `False` porque ninguna de las matrices dadas cumple las condiciones simétricas o estructurales estrictas sobre $\mathbb{C}$:

- **Matriz Hermitiana ($M = M^\dagger$)**: Exige que la matriz sea igual a su tranpuesta conjugada. Una propiedad fundamental es que **todos sus valores propios deben ser estrictamente números reales** ($\lambda \in \mathbb{R}$). En los resultados los valores propios tienen parte imaginaria no nula (ej. $\lambda_1 = -7.48 + 4.14i$ ), por lo que no pueden ser Hermitianas.
- **Matriz Unitaria ($U^\dagger U = I$)**: Representa transformaciones que conservan la norma/probabilidad. Exige que todos sus valores propios tengan módulo igual a 1 ($\vert\lambda\vert = 1$). En los resultados las matrices tienen módulos mayores a 1.
- **Matriz Normal ($M M^\dagger = M^\dagger M$)**: Requiere que la matriz conmute con su adjunta. Las matrices $A,B,C,D$ del ejercicio son matrices complejas genéricas no estructuradas.

## Vectores estados $\mathbb{C}^2: \vert{0}\rangle, \vert{1}\rangle$

En física cuántica, los estados de los sistemas no se representan con variables continuas ordinarias, sino con **vectores estado** en un espacio vectorial complejo ($\mathbb{C}^n$), utilizando la notacion _Dirac_ o _bra-ket($\vert\psi\rangle$)_.

Un qubit es el sistema cuántico de 2 niveles. Sus estados base son

$$\vert{0}\rangle = \begin{bmatrix} 1 \\ 0 \end{bmatrix}, \quad \vert{1}\rangle = \begin{bmatrix} 0 \\ 1 \end{bmatrix}$$

Cualquier estado generico $\vert\psi\rangle$ es una superposicion: 

$$\vert\psi\rangle = \alpha \vert{0}\rangle + \beta \vert{1}\rangle = \begin{bmatrix} \alpha \\ \beta \end{bmatrix}$$

donde $\alpha, \beta \in \mathbb{C}$ y cumplen la **condición de normalización**: $\vert\alpha\vert^2 + \vert\beta\vert^2 = 1$.

La constante de normalización $\frac{1}{\sqrt{2}}$ presente en los estados $\vert{+}\rangle,\vert{-}\rangle$, estados de Bell y los estados de 3 qubits provienen del **Axioma de Born** de la mecánica cuántica: La suma de las probabilidades de todos los resultados posobles debe ser exactamente 1 (100%).

$$\vert{\psi}\vert = \alpha\vert{000}\vert + \beta\vert{111}\vert \implies \vert{\alpha}\vert^2 + \vert{\beta}\vert^2 = 1$$

$$\left\vert\frac{1}{\sqrt{2}}\right\vert^2 + \left\vert\frac{1}{\sqrt{2}}\right\vert^2 = \frac{1}{2} + \frac{1}{2} = 1$$

#### Fuentes
- **Strang, G. (2016)**. Introduction to Linear Algebra (5th ed.). Wellesley-Cambridge Press. (Capítulo 4).
- **Qiskit Texbook**: Single Quibit Gates & States. IBM Quantum Experience

```python
from qiskit.quantum_info import Statevector

ket_0 = np.array([1,0], dtype=complex)
ket_1 = np.array([0,1], dtype=complex)

# Creacion de estados de superposicion
ket_suma_np = (1 / np.sqrt(2)) * (ket_0 + ket_1)

vector_estado_suma = Statevector(ket_suma_np)

norma_sq = np.vdot(ket_suma_np, ket_suma_np).real

display(Math(rf"""
\begin{{aligned}}
\vert{{0}}\rangle &= {formato_vector(ket_0)} \quad , \quad \vert{{1}}\rangle \text{{ = }} {formato_vector(ket_1)} \\
\vert{{+}}\rangle &= \frac{{1}}{{\sqrt{{2}}}}(\vert{{0}}\rangle + \vert{{1}}\rangle) \text{{ = }} {formato_vector(ket_suma_np)} \\
\text{{Comprobación de la norma }} \\
\langle +\vert{{+}}\rangle &= {norma_sq:.2f} \implies \text{{ Estado valido normalizado }}
\end{{aligned}}
"""))
```

## Estados $\vert{+}\rangle, \vert-\rangle, \vert{}i+\rangle, \vert{}i-\rangle$ y estados de Bell

En computación cuantica, ademas de la base computacional $\{\vert{}0\rangle, \vert{}1\rangle\}$, usamos frecuentemente otras bases ortonormales en $\mathbb{C}^2$

### **Base X ($\vert{+}\rangle,\vert{-}\rangle$)**
- $\vert{+}\rangle = \frac{1}{\sqrt{2}}(\vert{0}\rangle + \vert{1}\rangle)$
- $\vert{-}\rangle = \frac{1}{\sqrt{2}}(\vert{0}\rangle - \vert{1}\rangle)$

### **Base Y($\vert{i}+\rangle,\vert{i}-\rangle \text{ o } \vert{}+i\rangle, \vert{}-i\rangle$)**
- $\vert{i}+\rangle = \frac{1}{\sqrt{2}}(\vert{0}\rangle + i\vert{1}\rangle)$
- $\vert{i}-\rangle = \frac{1}{\sqrt{2}}(\vert{0}\rangle - i\vert{1}\rangle)$

### **Estados de Bell (Base de entrelazamiento en $\mathbb{C}^4$)**
Para representar un sistema de 2 qubits, combinamos los espacios mediante el producto de Kronecker/Tensorial:

$$\vert\psi_1\rangle \otimes \vert\psi_2\rangle = \begin{bmatrix} \alpha_1 \\ \beta_1 \end{bmatrix} \otimes \begin{bmatrix} \alpha_2 \\ \beta_2 \end{bmatrix} = \begin{bmatrix} \alpha_1 \alpha_2 \\ \alpha_1 \beta_2 \\ \beta_1 \alpha_2 \\ \beta_1 \beta_2 \end{bmatrix}$$

Los **estados de Bell** son los 4 estados maximamente entrelazados fundamentales en la información cuántica

- $\vert\Phi^+\rangle = \frac{1}{\sqrt{2}}(\vert{00}\rangle + \vert{11}\rangle) = \frac{1}{\sqrt{2}}\begin{bmatrix} 1 \\ 0 \\ 0 \\ 1 \end{bmatrix}$
- $\vert\Phi^-\rangle = \frac{1}{\sqrt{2}}(\vert{00}\rangle - \vert{11}\rangle) = \frac{1}{\sqrt{2}}\begin{bmatrix} 1 \\ 0 \\ 0 \\ -1 \end{bmatrix}$
- $\vert\Psi^+\rangle = \frac{1}{\sqrt{2}}(\vert{01}\rangle + \vert{10}\rangle) = \frac{1}{\sqrt{2}}\begin{bmatrix} 0 \\ 1 \\ 1 \\ 0 \end{bmatrix}$
- $\vert\Psi^-\rangle = \frac{1}{\sqrt{2}}(\vert{01}\rangle - \vert{10}\rangle) = \frac{1}{\sqrt{2}}\begin{bmatrix} 0 \\ 1 \\ -1 \\ 0 \end{bmatrix}$

#### Fuentes
- **Rieffel, E.G., & Polak, W.H. (2011)**. Quantum Computing: A Gentle Introduction. MIT Press(Capitulo 2, Capitulo 5)
- **Qiskit Texbook**: Single Quibit Gates & States. IBM Quantum Experience

```python
# En python para el calculo tensorial se utiliza la funcion kron de numPy
ket_0 = np.array([1,0], dtype=complex)
ket_1 = np.array([0,1], dtype=complex)

# Estados de Qubit
base_constante = (1/np.sqrt(2))
ket_plus = base_constante * (ket_0 + ket_1)
ket_minus = base_constante * (ket_0 - ket_1)
ket_i_plus = base_constante * (ket_0 + 1j * ket_1)
ket_i_minus = base_constante * (ket_0 - 1j * ket_1)
prod_plus_minus = np.vdot(ket_plus, ket_minus)

# Productos tensoriales (|00>, |01>, |10>, |11>
ket_00 = np.kron(ket_0, ket_0)
ket_01 = np.kron(ket_0, ket_1)
ket_10 = np.kron(ket_1, ket_0)
ket_11 = np.kron(ket_1, ket_1)

# Construccion de los estados de Bell
bell_phi_plus = base_constante * (ket_00 + ket_11)
bell_phi_minus = base_constante * (ket_00 - ket_11)
bell_psi_plus = base_constante * (ket_01 + ket_10)
bell_psi_minus = base_constante * (ket_01 - ket_10)

estado_vector = Statevector(bell_phi_plus)
display(Math(rf"""
\begin{{aligned}}
\mathbf{{\text{{Estados de 1 Qubit (Bases X e Y)}}}} &: \\[0.3em]
\vert{{+}}\rangle &= {formato_vector(ket_plus)} \quad, \quad \vert{{-}}\rangle \text{{ = }} {formato_vector(ket_minus)} \\[0.5em]
\vert{{i+}}\rangle &= {formato_vector(ket_i_plus)} \quad, \quad \vert{{i-}}\rangle \text {{ = }} {formato_vector(ket_i_minus)} \\[0.5em]
\langle + \vert{{-}}\rangle &= {formato_binomial(prod_plus_minus)} \\[1.5em]
\mathbf{{\text{{Estados de Bell (Entrelazamiento en }} \mathbb{{C}}^4\mathbf{{)}}}} &:\\[0.3em]
\vert{{\Phi^+}}\rangle &= \frac{{1}}{{\sqrt{{2}}}}(\vert{{00}}\rangle + \vert{{11}}\rangle) \text{{ = }} {formato_vector(bell_phi_plus)} \\[0.5em]
\vert{{\Phi^-}}\rangle &= \frac{{1}}{{\sqrt{{2}}}}(\vert{{00}}\rangle - \vert{{11}}\rangle) \text{{ = }} {formato_vector(bell_phi_minus)} \\[0.5em]
\vert{{\Psi^+}}\rangle &= \frac{{1}}{{\sqrt{{2}}}}(\vert{{01}}\rangle + \vert{{10}}\rangle) \text{{ = }} {formato_vector(bell_psi_plus)} \\[0.5em]
\vert{{\Psi^-}}\rangle &= \frac{{1}}{{\sqrt{{2}}}}(\vert{{01}}\rangle - \vert{{10}}\rangle) \text{{ = }} {formato_vector(bell_psi_minus)}
\end{{aligned}}
"""))
```

## A partir de los vectores estado creados, encuentra la base computacional de $\mathbb{C}^4$
Para un sistema de 2 qubits, el espacio de estados es $\mathbb{C}^4 (\mathbb{C}^2 \otimes \mathbb{C}^2)$

La base de computacion de $\mathbb{C}^4$ es el conjunto de los 4 estados canónicos (o estados puros) sin superposición, formados por el producto tensorial ($\otimes$) de la base de 1 qubit $\{\vert{}0\rangle, \vert{}1\rangle\}$:
$$
\mathcal{B}_{\mathbb{C}^4} = \{\vert{00}\rangle, \vert{01}\rangle, \vert{10}\rangle, \vert{11}\rangle\}
$$

Donde el orden canónico en álgebra de matriz columna es:
- $\vert{00}\rangle = \vert{0}\rangle \otimes \vert{0}\rangle:$ Representa el estado $0$ en binario ($00_2 = 0$).
- $\vert{01}\rangle = \vert{0}\rangle \otimes \vert{1}\rangle:$ Representa el estado $1$ en binario ($01_2 = 1$).
- $\vert{10}\rangle = \vert{1}\rangle \otimes \vert{0}\rangle:$ Representa el estado $2$ en binario ($10_2 = 2$).
- $\vert{11}\rangle = \vert{1}\rangle \otimes \vert{1}\rangle:$ Representa el estado $3$ en binario ($11_2 = 3$).

Para que un conjunto de vectores $\vert{}\psi_i\rangle$ sea **ortonormal**, debe cumplir con 2 cosas a través del **producto interno** (producto escalar complejo $\langle\psi_i\vert{\psi_j}\rangle$):
1. **Ortogonalidad**: El producto interno entre 2 vectores diferentes debe ser exactamente $0(\langle\psi_i\vert{\psi_j}\rangle = 0 \text{ si } i \neq j )$. No interfieren entre si.
2. **Normalización**: El producto interno de un vector consigo mismo debe ser exactamente 1 ($\langle\psi_i\vert{\psi_j\rangle} = 1$). La probabilidad total de encontrar el sistema suma 100%

En notación matemática, esto se resume con la delta de **Kronecker** $\langle \phi_i \vert{\phi_j} \rangle = \delta_{ij}$

#### Fuentes
- **Strang, G. (2016)**. Introduction to Linear Algebra (5th ed.). Wellesley-Cambridge Press. (Capítulo 4).
- **Rieffel, E.G., & Polak, W.H. (2011)**. Quantum Computing: A Gentle Introduction. MIT Press(Capitulo 2)
- **Qiskit Texbook**: Single Quibit Gates & States. IBM Quantum Experience

```python
# En python el producto interno entre dos cectores complejos se calcula con la funcion np.vdor(phi, psi)

base_c4 = [ket_00, ket_01, ket_10, ket_11]
# Construimos la matriz de GRAM
matriz_gram = np.zeros((4,4), dtype=complex)

for i in range(4):
    for j in range(4):
        matriz_gram[i,j] = np.vdot(base_c4[i], base_c4[j])

es_ortonormal = np.allclose(matriz_gram, np.eye(4))

display(Math(rf"""
\begin{{aligned}}
\text{{Matriz de productos Internos}} \langle \phi_i\vert{{\phi_j}}\rangle &: \\
M_{{\text{{Gram}}}} &= {formato_matriz(matriz_gram)} \\[0.8em]
\text{{Es la matriz Identidad?}} &\implies \mathbf{{{es_ortonormal}}} \quad (\delta{{ij}}) \\
\therefore \text{{La base }} &\{{\vert{{00}}\rangle, \vert{{01}}\rangle, \vert{{10}}\rangle, \vert{{11}}\rangle\}} \text{{ es estrictamente }} \mathbf{{\text{{ORTONORMAL}}}}
\end{{aligned}}
"""))
```

## Sistema de 3 Qubits

Para trabajar con 3 qubits como $\vert{000}\rangle$ o $\vert{111}\rangle$, el espacio de estos pasa a ser $\mathbb{C}^8 (\mathbb{C}^2 \otimes \mathbb{C}^2 \otimes \mathbb{C}^2)$.

Los 2 vectores canónicos base de 3 qubits que necesitamos son:
1. $\vert{000}\rangle = \vert{0}\rangle \otimes \vert{0}\rangle \otimes \vert{0}\rangle$: Un vector columna de 8 elementos con un 1 en la primera posición.
2. $\vert{111}\rangle = \vert{1}\rangle \otimes \vert{1}\rangle \otimes \vert{1}\rangle$: Un vector columna de 8 elementos con un 1 en la última posición.

### Encontrar los siguientes vectores
1. $\frac{1}{\sqrt{2}}\vert{000}\rangle + \frac{1}{\sqrt{2}}\vert{111}\rangle$
2. $\frac{i}{\sqrt{2}}\vert{000}\rangle - \frac{1}{\sqrt{2}}\vert{111}\rangle$
3. $\frac{1}{\sqrt{2}}\vert{000}\rangle + \frac{i}{\sqrt{2}}\vert{111}\rangle$

- **Qiskit Texbook**: Single Quibit Gates & States. IBM Quantum Experience
